## Project Overview

In this notebook, we’ll scrape match data from the [WorldFootball.net Premier League 2024–25 Matches page](https://www.worldfootball.net/all_matches/eng-premier-league-2024-2025/). The table contains over 380 rows of rich, structured data, including:

- **Match Date**  
- **Home Team** & **Away Team**  
- **Final Score**  
- **Attendance** figures  
- **Venue** names  

This dataset will allow us to visualize trends such as attendance over time, goal distributions, and team performance.


## 1. Import Libraries  


```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## 2. Fetch the Page  


```python
url = "https://www.worldfootball.net/all_matches/eng-premier-league-2024-2025/"
response = requests.get(url)
response.raise_for_status()   # Ensure we got a 200 OK
```

## 3. Parse HTML and Locate Table  


```python
soup = BeautifulSoup(response.text, "html.parser")
table = soup.find("table", {"class": "standard_tabelle"})
```

## 4. Convert to pandas DataFrame  


```python
# 4. Convert HTML Table to DataFrame (future-proofed)
from io import StringIO

html_str = str(table)                         # your BeautifulSoup table
df = pd.read_html(StringIO(html_str))[0]      # wrap the HTML in StringIO
```

```python
df.to_csv("CSC221-webscrape-data.csv", index=False)
print("✅ Data saved to CSC221-webscrape-data.csv")
```
